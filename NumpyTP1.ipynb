# punto 7 Crear dos arrays A=[1,2,3,4,5] y B=[10,20,30,40,50]. Calcular suma, resta, multiplicación y división elemento a elemento.

```python
A=[1,2,3,4,5]
B=[10,20,30,40,50]
arrayA= np.array(A)
arrayB = np.array(B)

suma = arrayA + arrayB
resta = arrayA - arrayB
multiplicacion = arrayA * arrayB
division = arrayA / arrayB


print("Suma:", suma)
print("Resta:", resta)
print("Multiplicación:", multiplicacion)
print("División:", division)
```

# 9. Crear una matriz 4x4 con números del 1 al 16. Extraer la segunda fila, la tercera columna y el elemento (2,2).

```python
matriz = np.arange(1, 17).reshape(4, 4)
print("matriz", matriz)

segunda_fila = matriz[1, :]
print("segunda fila", segunda_fila)

tercera_columna = matriz[:, 2]
print("tercera fila", tercera_columna)

elemento_2_2 = matriz[2, 2]
print("elemento (2,2)", elemento_2_2)
```

# 10. Crear una matriz 3x3 con números aleatorios entre 1 y 10. Calcular la suma de cada fila y de cada columna.

```python
m = np.random.randint(1,11, size=(3,3))
print("matriz", m)
#axis = 1 fila
suma_filas = np.sum(m, axis=1)
print("suma de cada fila", suma_filas)

#axis = 0 columna
suma_columnas = np.sum(m, axis=0)
print("Suma de cada columna", suma_columnas)
```

# 15. Dado el array [10,20,30,40,50], normalizar los valores usando (valor - minimo) / (maximo - minimo).

```python
array = np.array([10,20,30,40,50])
print(array)
array_normalizado = (array - array.min()) / (array.max() - array.min())
print("normalizado",array_normalizado)
```
